Low Level File Open

Flags in os.open

| Flag      | Meaning      |
| ------------- | ------------ |
| `os.O_RDONLY` | Read only    |
| `os.O_WRONLY` | Write only   |
| `os.O_RDWR`   | Read & Write |
| `os.O_CREAT`  | Create file  |
| `os.O_APPEND` | Append mode  |


```python
# Low-Level File open
fd = os.open("Demo.txt",os.O_RDWR, 0o600)
os.write(fd,b"Using Low-Level Opening\n")
os.lseek(fd,0,0)
k = os.read(fd,200)
# k = k.split("\n") as the os.read() returns bytes we cannot use split directly
k = k.decode().split("\n")
for i in k:
    print(i)
os.write(fd,b"\nUsing Low-Level Opening\n")
os.lseek(fd,0,0)
k = os.read(fd,200)
k = k.decode().split("\n")
for i in k:
    print(i)
os.close(fd)
```

Valid whence Values

| Value | Constant      | Meaning               |
| ----- | ------------- | --------------------- |
| `0`   | `os.SEEK_SET` | From beginning        |
| `1`   | `os.SEEK_CUR` | From current position |
| `2`   | `os.SEEK_END` | From end of file      |


```python
f = os.open("Demo.txt",os.O_RDWR|os.O_APPEND,0o600)
os.write(f,b"\nAppending Using Os\n")
os.lseek(f,15,0)
print(os.read(f,500))
os.close(f)
```

```python
def FileOpener(fn,flag="r",data=b""):
    if flag == "w+" or flag=="w" or flag=="a" or flag=="r+" or flag=="a+":
        fd = os.open(fn,os.O_RDWR|os.O_APPEND,0o600)
        os.write(fd,data)
        os.lseek(fd,0,0)
        print(os.read(fd,1000))
        os.close(fd)
    else:
        fd = os.open(fn,os.O_RDWR,0o600)
        print(os.read(fd))
        os.close(fd)

FileOpener("Demo.txt","w+",b"Just Writing the data using Functions")
```
