```python
import pandas as pd
```

```python
s = pd.Series([1,2,3])
print(s)
df = pd.DataFrame({
    'A':[1,2,3],
    'B':[4,5,6],
    'C':[7,8,9]
})
print(df)
print(f"___Head___\n{df.head()}\n")
print(f"___Tail___\n{df.tail()}\n")
print(f"___Info___")
df.info()
print(f"__Describe__\n{df.describe()}")
```

```python
print(df['A'])
print(df['B'])
print(df.loc[1])
print(df.loc[1,'A'])
print(df.iloc[1,0])
print(df.iloc[1])
```

```python
print(df[df['C']>7])
print(df['C'])
df['C'] = df['A']+df['B']
print(df['C'])
df = df.drop('C',axis=1)
print(df)
print(df.sort_values("A"))
```

```python
df2 = pd.DataFrame({
    "team":["A","B","C","D"],
    "Scores":[22,55,33,45]
})
print(df2)
print(df2.groupby("team").sum())
```

```python
lf = pd.DataFrame({
    "id":[1,2],
    "Val":[10,20]
})
rt = pd.DataFrame({
    'id':[1,2],
    "Val2":[100,200]
})
print(pd.merge(lf,rt,on="id"))
```

```python
df3 = pd.DataFrame({"A":[1,2,3]})
print(df3["A"].apply(lambda x:x*5))
df3 = pd.DataFrame({"char":['a','b','c']})
print(df3['char'].str.upper())
```

```python
dates = pd.date_range("2024-01-01",periods=3)
df = pd.DataFrame({'dates':dates})
print(df['dates'].dt.year)
print(df['dates'].dt.month)
print(df['dates'].dt.day)
print(df['dates'].dt.weekday)
print(df['dates'].dt.day_name())
print(df)
```
